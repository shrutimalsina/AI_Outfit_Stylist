# Dataset exploration

Polyvore outfits are human-curated groups. Ingestion samples complete outfits, preserves official split membership, and records validation exclusions.


```python
import os, sys
from pathlib import Path
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'ARCHITECTURE.md').is_file())
sys.path.insert(0, str(root))
from pipeline.prepare_dataset import read_json
fixture = root / 'data/processed/fixture'
manifest_path = Path(os.environ.get('STYLIST_MANIFEST', fixture / 'manifest.json'))
manifest = read_json(manifest_path)
image_root = Path(os.environ.get('STYLIST_IMAGE_ROOT', fixture))
print('Dataset:', manifest['dataset'], '| Revision:', manifest['revision'])
```

```python
from collections import Counter
import matplotlib.pyplot as plt
from PIL import Image
print('Outfits by split:', Counter(outfit['split'] for outfit in manifest['outfits']))
print('Cleaning exclusions:', manifest.get('exclusion_counts', {}))
counts = Counter(item['category'] for item in manifest['items'])
plt.bar(list(counts), list(counts.values()))
plt.xticks(rotation=30)
plt.ylabel('Unique items')
plt.show()
example = manifest['outfits'][0]
items = {item['id']: item for item in manifest['items']}
fig, axes = plt.subplots(1, len(example['item_ids']), squeeze=False)
for ax, item_id in zip(axes.flat, example['item_ids']):
    item = items[item_id]
    with Image.open(image_root / item['image']) as image:
        ax.imshow(image)
    ax.set_title(item['category'])
    ax.axis('off')
plt.show()
```
